```python
import ast
import pandas as pd

main_dir = '/home/gap2/Documents/PPRL/PPRL-Dev-main/results/blocking/D3/'
```

```python
def process_df (bits):
    df = pd.read_csv(main_dir + 'blocking_encoder_' + str(bits) + '.csv')
    df['bits'] = bits
    df['RR'] = 1-df['candidate_pairs']/(1354 * 3039) # Lefteri, check whether this is correct
    df['config_dict'] = df['config'].apply(ast.literal_eval)
    config_cols = pd.json_normalize(df['config_dict'])
    df = df.join(config_cols)
    df = df.drop(columns=['config', 'config_dict'])
    
    return df[((df['RR']>0.25) & (df['Recall %']>=90))].copy()
```

```python
df1 = process_df(1)
df1
```

```python
df2 = process_df(2)
df2
```

```python
df4 = process_df(4)
df4
```

```python
df8 = process_df(8)
df8
```

```python
df10 = process_df(10)
df10
```

```python
df12 = process_df(12)
df12
```

```python
result = pd.DataFrame()
dfs = [df1, df2, df4, df8, df10, df12]
for df in dfs:
    result = pd.concat([result, df], ignore_index=True, sort=False)
result
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=result,
    x="size",
    y="Recall %",
    hue="bits"
)

plt.title("Recall grouped by Size and Bits")
plt.xlabel("Size")
plt.ylabel("Recall")
plt.legend(title="Bits")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=result,
    x="size",
    y="RR",
    hue="bits"
)

plt.title("Reduction Ratio grouped by Size and Bits")
plt.xlabel("Size")
plt.ylabel("RR")
plt.legend(title="Bits")
plt.tight_layout()
plt.show()
```
