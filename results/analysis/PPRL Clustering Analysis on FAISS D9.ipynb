```python
import ast
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

datasets = [ 'cartesian', 'flat FAISS', 'hnsw FAISS', 'lsh FAISS']
directories = [ '/home/gap2/Downloads/PPRL-Dev-main/results/clustering/D9/',
                '/home/gap2/Downloads/PPRL-Dev-main/results/faiss/D9/clustering/flat/',
                '/home/gap2/Downloads/PPRL-Dev-main/results/faiss/D9/clustering/hnsw/',
                '/home/gap2/Downloads/PPRL-Dev-main/results/faiss/D9/clustering/lsh/' ]
```

```python
def show_diagram(clustering_method, measure, label):
    dfs = []
    for i in range(len(datasets)):
        for sim in ["cosine", "dice", "jaccard"]:
            df = pd.read_csv(directories[i] + str(clustering_method)+f"_{sim}.csv", on_bad_lines="skip")
            df["similarity"] = sim
            df["dataset"] = datasets[i]
            dfs.append(df)

    result = pd.concat(dfs, ignore_index=True)
    result['config_dict'] = result['config'].apply(
        lambda x: ast.literal_eval(x) if pd.notna(x) else {}
    )
    config_cols = pd.json_normalize(result['config_dict'])
    result = result.join(config_cols)
    result = result.drop(columns=['config', 'config_dict'])
    
    max_f1 = result.loc[result.groupby(['num_hashes','dataset', 'similarity'])['F1 %'].idxmax()]
    
    g = sns.relplot(
        data=max_f1,
        x="num_hashes",
        y=measure,
        hue="dataset",
        style="dataset",
        markers=True,
        dashes=False,      # solid lines but different markers
        col="similarity",
        kind="line",
        linewidth=2.5,
        markersize=8
    )

    for ax in g.axes.flat:
        ax.tick_params(axis='both', labelsize=20)

    g.set_axis_labels("Number of Hashes", label, fontsize=20)
    g.set(xlim=(0, None), ylim=(0, 100))
    g.set_titles("{col_name}", size=20)

    plt.tight_layout()
    plt.show()
```

```python
show_diagram("KiralyMSMApproximateClustering", "F1 %", "F1 (%)")
```
